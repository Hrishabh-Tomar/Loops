1. Print 1 to 100

```python
# Initialization
i = 1
# Condition: keep going while i is 100 or less
while i <= 100:
    print(i)
    # Update: move toward termination
    i += 1
```

2. Print 100 to 1

```python
i = 100                  # Initialization
while i >= 1:            # Condition
    print(i)
    i -= 1               # Update (counts down, so the loop ends)
```

3. Even numbers between 1 and 100

```python
i = 2                    # Initialization: first even number
while i <= 100:          # Condition
    print(i)
    i += 2               # Update: jump by 2
```

4. Sum of digits

```python
n = int(input("Enter a number: "))

num = abs(n)             # Initialization: work on a positive copy
total = 0                # Running sum

while num > 0:           # Condition: digits remain
    total += num % 10    # Take the last digit
    num //= 10           # Update: drop the last digit (reaches 0 = termination)

print(f"Sum of {n} digits:", total)   
```

5. Reverse an integer

```python
n = int(input("Enter Your number: "))

sign = -1 if n < 0 else 1
num = abs(n)
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10
    
print(f"reversed number {n} is {sign * reverse}")
```

6. Count digits

```python
n = abs(int(input("Enter your Number: ")))

if n == 0:
    count = 1
else:
    count = 0
    while n > 0:
        count += 1
        n //= 10
        
print(f"count of digits: {count}")
```

7. Factorial

```python
n = int(input("Enter Your Number: "))

if n < 0:
    print("Factorial of negative is not defined")
else:
    result = 1
    i = n
    while i > 0:
        result *= i
        i -= 1
    print(f"{n} = {result}")
```

8. Keep reading numbers until 0, then show the sum

```python
total = 0
number = int(input("Enter A Number (0 TO STOP!!): "))

while number != 0:
    total += number
    number = int(input("Enter A number(Enter 0 to stop!!)  "))
    
print("total sum of the entered numbers: ", total)
```

9. Password checker

```python
CORRECT_PASSWORD = "Super30@123"

attempt = input("Enter password: ")           # Initialization: first attempt

# Condition: repeat until the password matches
# (intentionally runs until correct, not an accidental infinite loop)
while attempt != CORRECT_PASSWORD:
    print("Wrong password. Try again.")
    attempt = input("Enter password: ")        # Update: new attempt

print("Access granted!")
```

10. Guessing game

```python
secret_number = 34
attempt = 0
guess = int(input("Guess the number: "))
attempt += 1

while guess != secret_number:
    if guess < secret_number:
        print("Too Low, Guess Again")
    else:
        print("Too High, guess Again")
    guess = int(input("Guess the number: "))
    attempt += 1
        
print(f"Correct, you took {attempt} attempts to guess the number")
```

11. Menu-driven calculator

```python
choice = ""                                    # Initialization

while choice != "5":                           # Condition: run until Exit
    print("\n--- Calculator ---")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")
    choice = input("Enter your choice (1-5): ").strip()   # Update

    if choice == "5":                          # Termination
        print("Goodbye!")
    elif choice in ("1", "2", "3", "4"):
        try:
            a = float(input("Enter first number: "))
            b = float(input("Enter second number: "))
        except ValueError:
            print("Invalid number. Please try again.")
            continue

        if choice == "1":
            print("Result:", a + b)
        elif choice == "2":
            print("Result:", a - b)
        elif choice == "3":
            print("Result:", a * b)
        else:
            if b == 0:
                print("Error: cannot divide by zero.")
            else:
                print("Result:", a / b)
    else:
        print("Invalid choice. Please select 1-5.")
```

12. ATM menu

```python
balance = 10000.0                              # Initialization
running = True                                 # Loop control flag

while running:                                 # Condition: until user exits
    print("\n--- ATM ---")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")
    choice = input("Enter your choice (1-4): ").strip()

    if choice == "1":
        print(f"Current balance: {balance:.2f}")

    elif choice == "2":
        try:
            amount = float(input("Enter deposit amount: "))
        except ValueError:
            print("Invalid amount.")
            continue
        if amount > 0:
            balance += amount
            print(f"Deposited {amount:.2f}. New balance: {balance:.2f}")
        else:
            print("Amount must be greater than 0.")

    elif choice == "3":
        try:
            amount = float(input("Enter withdrawal amount: "))
        except ValueError:
            print("Invalid amount.")
            continue
        if amount <= 0:
            print("Amount must be greater than 0.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            print(f"Withdrew {amount:.2f}. New balance: {balance:.2f}")

    elif choice == "4":
        running = False                        # Termination: explicit Exit
        print("Thank you for using the ATM. Goodbye!")

    else:
        print("Invalid choice. Please select 1-4.")
```

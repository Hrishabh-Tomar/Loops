# Q1: Print numbers from 1 to 100 using a for loop

````python
for i in range(1,101):
    print(i)
````

# Q2: Print all even numbers from 1 to 100

````python
for num in range(1,101):
    if num % 2 == 0:
        print(num)
````

# Q3: Print all odd numbers from 1 to 100

````python
for num in range(1,101):
    if num % 2 != 0:
        print(num)
````

# Q4: Take an integer n and print its multiplication table from 1 to 20

````python
n = int(input("Enter the number: "))

for num in range(1,21):
    print(f"{n} * {num} -> {n * num}")
````

# Q5: Calculate the sum of numbers from 1 to n using a loop

````python
num = int(input("Enter the number: "))
total = 0
for i in range(1,num + 1):
    total += i
    
print(f"Sum of numbers from 1 to {num}: {total}")
````

# Q6: Calculate the factorial of a number without using any built-in factorial function

````python
n = int(input("Enter the Number: "))

if n < 0:
    print("factorial does not work for Negative numbers")
else:
    result = 1
    
    for i in range(1, n + 1):
        result *= i
        
    print(f"Factorial of {n} = {result} ")
````

# Q7: Print only the numbers divisible by 3

````python
numbers = [12, 7, 9, 20, 33, 42, 8, 15]

for num in numbers:
    if num % 3 == 0:
        print(num)
````

# Q8: Print every language along with its length

````python
languages = ["Python", "Java", "C++", "JavaScript", "Go"]

for i in languages:
    print(f" {i} -> {len(i)}")
````

# Q9: Iterate through the dictionary and print every key and value

````python
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Data Science",
    "city": "Bangalore"
}

for key, value in student.items():
    print(f"{key} : {value}")
````

# Q10: Count how many vowels exist in a user-provided string

````python
text = input("Enter your Text: ")

vowels = "aeiouAEIOU"
count = 0
for char in text:
    if char in vowels:
        count += 1
        
print(f"total vowels in text: {count}")
````

# Q11: Reverse a string using a for loop (no [::-1], no reversed())

````python
text = input("enter your string: ")

reverse_text = ""

for char in text:
    reverse_text = char + reverse_text

print(f"reverse text string is = {reverse_text}")
````

# Q12: Find the largest number from a list without using max()

````python
numbers = [12, 7, 9, 20, 33, 42, 8, 15]
largest = 0
for num in numbers:
    if num > largest:
        largest = num
        
print(f"largest number = {largest}")
````

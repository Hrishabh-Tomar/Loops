Q1: Print numbers from 1-100 but skip numbers divisible by 5 using continue.

````python
for num in range(1,101):
    if num % 5 == 0:
        continue
    print(num)
````

Q2: Iterate from 1-100 and stop when you encounter the first number
divisible by both 7 and 11, using break.

````python
for num in range(1,101):
    if num % 7 == 0 and num % 11 == 0:
        print("first number dividible by 7 and 11 both:",num)
        break
    print(num)
````

Q3: Search for a user-provided number inside a list using for-else.

````python
numbers = [12, 45, 7, 89, 34, 23, 67, 90, 15, 3]

target = int(input("enter number to find: "))

for num in numbers:
    if num == target:
        print("you found the number")
        break
else:
    print("number not found")
````

Q4: Use enumerate() to display each name with its 1-based position.

````python
names = ["Aman", "Ravi", "Sudhanshu", "Priya", "Anjali"]
for index, name in enumerate(names, start=1):
    print(index, name)
````

Q5: Print an ascending star pattern:

````python
rows = 5
for i in range(1, rows + 1):
    print("*" * i)
````

Q6: Print a descending star pattern:

````python
rows = 5
for i in range(rows,0,-1):
    print("*" * i)
````

Q7: Generate multiplication tables from 1 to 10 using nested loops.

````python
for table in range(1,11):
    for i in range(1,11):
        print(f"{table} * {i} -> {table * i}")
        
    print()
````

Q8: Find all numbers between 1 and 200 divisible by both 3 and 5.

````python
result = []
for num in range(1,201):
    if num % 3 == 0 and num % 5 == 0:
        result.append(num)
print(f"number divisible by 3 and 5: {result}")
````

Q9: Given a list containing duplicate elements, create another list
containing only unique elements without using set()

````python
numbers = [1, 2, 3, 2, 4, 5, 1, 6, 3, 7, 5, 8]

unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)
    
print("unique list: ", unique)
print("original list: ", numbers)
````

Q10: Given a list of numbers, count positive numbers, negative numbers,
and zeros.

````python
numbers = [10, -4, 8, -2, 0, 15, -9, 21]

positive_numbers = 0
neg_numbers = 0
zero_numbers = 0

for num in numbers:
    if num > 0:
        positive_numbers += 1
    elif num < 0:
        neg_numbers += 1
    else:
        zero_numbers += 1
        
print("postive numbers in alist : ", positive_numbers)
print("negative numbers in a list: ", neg_numbers)
print("zeros in a list: ", zero_numbers)
    
````

Q11: Determine whether a number is prime using a loop.

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2,int(n ** 0.5) + 1):
        if n % i == 0:
            return False
        
    return True

num = int(input("Enter your Number: "))

if is_prime(num):
    print(f"{num} is prime number")
else:
    print(f"{num} is not a prime number")
````

Q12: Print all prime numbers between 1 and 100.

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5 + 1)):
        if n % i == 0:
            return False
    return True

Primes = [num for num in range(2,101) if is_prime(num)]

print("Prime number from 1 to 100: ")
print(Primes)
        
````

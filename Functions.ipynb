"Q2: Create functions for addition, subtraction, multiplication, division."""

```python
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mult(a, b):
    return a * b

def div(a, b):
    if b == 0:
        raise ValueError
    return a / b


if __name__ == "__main__":
    print("add(10, 5) =", add(10, 5))
    print("sub(10, 5) =", sub(10, 5))
    print("mult(10, 5) =", mult(10, 5))
    print("div(10, 5)   =", div(10, 5))
    try:
        div(10, 0)
    except ValueError as error:
        print("divide(10, 0)   -> Error:", error)
        
```

"""Q3: Create a function that determines whether a number is even or odd."""


```python
def is_even(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"
    
for i in range(1,11):
    print(f"{is_even(i)} -> {i}")
    
```

"""Q4: Create a function that returns the largest of three numbers without max()."""

```python
def largest_number(a,b,c):
    largest = a
    if b > largest:
        largest = b
    if c > largest:
        largest = c
    
    return largest

print(largest_number(12,54,67))
```

"""Q5: Create a function that calculates factorial."""


```python
def factorial(n):
    if n < 0:
        raise ValueError
    result = 1
    for i in range(2, n + 1):
        result *= i
    
    return result

print(factorial(3))
print(factorial(5))
```

"""Q6: Create a function that checks whether a number is prime."""


```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
  

num = [1,3,5,6,78,7,9,0,0,90,7,0]
for n in num:
    print(f"{n} is prime? {is_prime(n)}")
```

"""Q7: Create def calculate_discount(price, discount=10) - default discount is 10%."""

```python
def discounted_price(price, discount=10):
    discount_amount = price * discount / 100
    return price - discount_amount

print(discounted_price(2000))
print(discounted_price(3400, 24))
    
```

"""Q8: Create a function that accepts a list and returns its sum without sum()."""

```python
def sum(numbers):
    total = 0
    for n in numbers:
        total += n
    return total

print(sum([1,4,66,7,88,9]))
```

""Q9: Create a function that accepts a string and returns the number of vowels."""

```python
def vowels(text):
    vow = "aeiouAEIOU"
    count = 0
    for i in text:
        if i in vow:
            count += 1
    return count

print(vowels("sdhbcshdbca"))
            
```

"""Q10: Create a function that accepts a string and checks whether it is a palindrome."""

```python
def is_palindrome(text):
    text = text.lower()
    reversed = ""
    for char in text:
        reversed = char + reversed
    return text == reversed

print(is_palindrome("madam"))
print(is_palindrome("matata"))
```

"""Q12: Demonstrate the difference between a function that uses print() and one
that uses return. Explain through code why return is useful."""

```python

def add_with_print(a, b):
    """Print the sum of two numbers. Nothing is given back to the caller."""
    print("Sum is:", a + b)
    # print() only SHOWS the answer on the screen.
    # There is no return statement, so Python gives back None by default.


def add_with_return(a, b):
    """Return the sum of two numbers so the caller can use it."""
    return a + b
    # return HANDS the answer back to the line that called the function.
    # It shows nothing on the screen by itself.


# ---------- Function with print() ----------
print("--- Function with print() ---")
result1 = add_with_print(5, 3)     # screen shows: Sum is: 8
print("Value stored:", result1)    # screen shows: Value stored: None

# Why are they different?
# "Sum is: 8" was displayed by the print() INSIDE the function.
# But the function never returned anything, so result1 holds None.
# The 8 appeared on the screen only; it was never saved in result1.

try:
    print(result1 * 2)             # trying to reuse the value
except TypeError as error:
    print("Cannot reuse it ->", error)


# ---------- Function with return ----------
print("\n--- Function with return ---")
result2 = add_with_return(5, 3)    # screen shows nothing yet
print("Value stored:", result2)    # screen shows: Value stored: 8

# Here the 8 was returned, so it is stored in result2.
# Now we can reuse it however we want:
print("Double of it:", result2 * 2)
print("Plus 10:", result2 + 10)
```
